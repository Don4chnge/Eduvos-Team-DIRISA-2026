# Statement of the Problem

## Initial problem statement (before EDA)

*Written at the start of the project. The ward-level analysis and voter registration data it mentions were planned but, as the analysis developed, not used; see the final problem statement below.*

South Africa's local government elections do not fail the same way in every municipality, and we want to understand why. This project investigates whether voter turnout follows a predictable, non-linear pattern shaped by political competition: an inverted U-curve where turnout peaks in competitive "battleground" municipalities and drops off at both extremes, in towns dominated by a single party for years where voters feel their vote is a formality, and in municipalities fractured across many small parties where unstable coalitions wear voters down. This matters to the Electoral Commission of South Africa, civil society groups trying to re-engage disillusioned communities, and journalists and researchers covering the 2026 elections, all of whom need a data backed explanation rather than anecdote. We approach this at the ward and municipal level, since that is where these dynamics actually play out, using historical local government election results from 2000 onward, extended to more recent cycles where municipal level data allows, combined with current voter registration data ahead of the 4 November 2026 Local Government Elections. Using vote share data, we will compute a party fragmentation index and a turnout rate for each municipality per election cycle, then test whether the relationship between them forms the inverted U pattern we expect. We acknowledge this model captures only part of the picture, since corruption, service delivery failures, and candidate quality also shape turnout, but competition driven apathy and fragmentation driven fatigue are two structural forces we can actually measure with the data available to us.

## What the EDA showed

We tested the inverted-U hypothesis across all five local elections (2000–2021) using the Lind–Mehlum U-test, and within the same municipality over time using fixed-effects models. **It was not supported:** the inverted U did not hold in any election, and fragmentation explained less than 1% of a municipality's changes in turnout. Turnout is driven mainly by national swings between elections and stable regional differences.

What the data showed clearly was the **collapse of one-party dominance**: municipalities dominated by one party fell from 70% in 2000 to 16% in 2021, and hung councils rose from 27 after 2016 to 72 after 2021. This led us to the final problem statement.

## Final problem statement (after EDA)

South Africa's local politics has changed dramatically since 2000. One-party dominance, once the norm in 70% of municipalities, had fallen to 16% by 2021, and hung councils, where no party wins a majority and a coalition must govern, jumped from 27 after the 2016 election to 72 after 2021. Coalition councils have been linked to unstable administrations, frequent changes of mayor and disrupted service delivery, yet there is no systematic, data-backed way to see where this is likely to happen next.

We began by asking whether voter turnout follows a non-linear, inverted U-shaped pattern driven by political competition: peaking in competitive "battleground" municipalities and dropping off both in one-party towns, where voters feel their vote is a formality, and in fragmented municipalities, where unstable coalitions wear voters down. Using official IEC results for every local government election from 2000 to 2021, we computed a fragmentation index (the Effective Number of Parties) and a turnout rate for each municipality in each election, and tested this relationship rigorously. The evidence did not support it: turnout is driven mainly by national swings between elections and stable regional differences, not by how competitive a municipality is.

What the data did show, clearly, was the collapse of one-party dominance itself. This led us to the question that matters most ahead of 4 November 2026: **which South African municipalities are most likely to experience a hung council after the 2026 Local Government Elections, and how can two decades of political fragmentation data be used to forecast where coalition government will be needed?** This matters to voters and journalists, who need to know where the result could change who governs; to political parties preparing for coalition negotiations; and to the Electoral Commission, government and civil society, who can prepare for instability in the councils most at risk rather than reacting after the results. We acknowledge that our model captures only part of the picture: polling, new parties, corruption, service delivery failures and candidate quality also shape outcomes, which is why our dashboard lets users test different election-day scenarios. Vote shares, fragmentation and their trends over five elections are the structural forces we can measure reliably with the data available to us.
